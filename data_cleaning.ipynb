```python
import pandas as pd
import numpy as np
df = pd.read_csv("retail_sales_dataset.csv")
df.head()
```

```python
print("Shape:", df.shape)
df.info()

df.describe(include="all")
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Gender"] = df["Gender"].str.strip().str.title()

df["Product Category"] = (
    df["Product Category"]
    .str.strip()
    .str.title()
)

df.info()

print(df["Gender"].unique())
print(df["Product Category"].unique())
```

```python
numeric_columns = [
    "Age",
    "Quantity",
    "Price per Unit",
    "Total Amount"
]
for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\nColumn: {col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", len(outliers))
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Gender"] = df["Gender"].fillna(
    df["Gender"].mode()[0]
)

df = df.sort_values("Date").reset_index(drop=True)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")

cleaned_df = pd.read_csv("cleaned_dataset.csv")

print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows",
        "Outliers"
    ],
    "Before Cleaning": [
        1000,
        9,
        0,
        0,
        0
    ],
    "After Cleaning": [
        len(df),
        len(df.columns),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        0
    ]
})

print(comparison)
```

```python
comparison.to_csv("data_quality_comparison.csv", index=False)

print("Data quality comparison saved successfully!")
```
