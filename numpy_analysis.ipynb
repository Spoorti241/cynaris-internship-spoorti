```python
import numpy as np
import time

print("NumPy Version:", np.__version__)
```

```python
arr1 = np.array([10, 20, 30, 40, 50])

arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:\n", arr1)
print("\n2D Array:\n", arr2)
print("\n3D Array:\n", arr3)

print("\nShapes:")
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
```

```python
a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])

print("Addition:", a + b)

print("Multiplication:", a * b)

print("Mean:", np.mean(a))

print("Sum:", np.sum(a))

print("Maximum:", np.max(a))
```

```python
marks = np.array([45, 78, 90, 32, 65, 88])

print("Marks above 60:", marks[marks > 60])

print("Marks below 50:", marks[marks < 50])

print("Marks between 60 and 90:", marks[(marks >= 60) & (marks <= 90)])

print("Selected positions:", marks[[0, 2, 4]])
```

```python
sales = np.array([120, 150, 180, 200, 250, 300, 350, 400])

print("Mean:", np.mean(sales))
print("Standard Deviation:", np.std(sales))
print("50th Percentile:", np.percentile(sales, 50))
print("90th Percentile:", np.percentile(sales, 90))

advertising = np.array([10, 20, 30, 40, 50])
revenue = np.array([100, 200, 300, 400, 500])

correlation = np.corrcoef(advertising, revenue)

print("Correlation:\n", correlation)
```

```python
size = 1_000_000

python_data = list(range(size))

start = time.time()
python_result = [x * 2 for x in python_data]
python_time = time.time() - start

numpy_data = np.arange(size)

start = time.time()
numpy_result = numpy_data * 2
numpy_time = time.time() - start

print("Python execution time:", python_time)
print("NumPy execution time:", numpy_time)

print("Python first 5 values:", python_result[:5])
print("NumPy first 5 values:", numpy_result[:5])

print("NumPy speedup:", python_time / numpy_time)
```
