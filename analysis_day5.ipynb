```python
import pandas as pd

data = {
    "Employee": ["Amit", "Priya", "Rahul", "Sneha", "Arjun", "Kavya", "Ravi", "Neha"],
    "Region": ["South", "North", "South", "East", "West", "North", "East", "South"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile"],
    "Sales": [50000, 30000, 20000, 45000, 35000, 25000, 55000, 40000],
    "Quantity": [2, 3, 1, 2, 4, 2, 3, 4]
}

df = pd.DataFrame(data)

df.to_csv("sales.csv", index=False)

print("Dataset created successfully!")
```

```python
category_sales = df.groupby("Product Category")["Total Amount"].sum()

print(category_sales)
```

```python
import matplotlib.pyplot as plt

category_sales.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
customer_transactions = df["Customer ID"].value_counts()

print(customer_transactions.head(10))
```

```python
gender_sales = df.groupby("Gender")["Total Amount"].sum()

print(gender_sales)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Month"] = df["Date"].dt.month_name()

monthly_sales = df.groupby("Month")["Total Amount"].sum()

print(monthly_sales)
```

```python
df["Month"] = df["Date"].dt.month

monthly_sales = df.groupby("Month")["Total Amount"].sum()

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
best_category = df.groupby("Product Category")["Quantity"].sum().idxmax()

print("Best-selling category:", best_category)
```
