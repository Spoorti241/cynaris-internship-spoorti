```python
import pandas as pd

employees = pd.DataFrame({
    'Employee_ID': [1, 2, 3, 4],
    'Name': ['Asha', 'Ravi', 'Priya', 'Kiran'],
    'Department_ID': [101, 102, 101, 103]
})

departments = pd.DataFrame({
    'Department_ID': [101, 102, 104],
    'Department': ['HR', 'IT', 'Finance']
})

print("Employees:")
display(employees)

print("Departments:")
display(departments)
```

```python
inner_result = pd.merge(
    employees,
    departments,
    on='Department_ID',
    how='inner'
)

display(inner_result)
```

```python
left_result = pd.merge(
    employees,
    departments,
    on='Department_ID',
    how='left'
)

display(left_result)
```

```python
right_result = pd.merge(
    employees,
    departments,
    on='Department_ID',
    how='right'
)

display(right_result)
```

```python
outer_result = pd.merge(
    employees,
    departments,
    on='Department_ID',
    how='outer'
)

display(outer_result)
```

```python
sales = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South', 'East', 'East'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 20000, 45000, 25000, 60000, 30000]
})

display(sales)
```

```python
pivot_result = pd.pivot_table(
    sales,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

display(pivot_result)
```

```python
wide_data = pd.DataFrame({
    'Region': ['North', 'South', 'East'],
    'Laptop': [50000, 45000, 60000],
    'Mobile': [20000, 25000, 30000]
})

display(wide_data)
```

```python
long_data = pd.melt(
    wide_data,
    id_vars=['Region'],
    var_name='Product',
    value_name='Sales'
)

display(long_data)
```

```python
result = (
    sales
    .query("Sales > 20000")
    .groupby('Region')['Sales']
    .sum()
    .sort_values(ascending=False)
)

display(result)
```
